```python
import statsmodels.api as sm

X = df[['Glucose', 'BMI', 'Age', 'Insulin']]
X = sm.add_constant(X)
y = df['Outcome']

modelo = sm.Logit(y, X).fit()
print(modelo.summary())
```

```python
# el modelo explica el 24% de probabilidad mediante el r¨2, además si aplicamos el odds ratio a
#l coeficiente de la glucose podemos observar OR=e0.034≈1.035, este resultado explica que aumenta la probabilidad
# de tener diabetes en un 3.3% al consumir glucose, tambien otros valores como el BMI con su coeficiente aumenta en un 8.8% sobre
#quienes tienen glucosa y cada año adicional de edad aumenta el riesgo en ~2.9%
```

```python
from scipy.stats import ttest_ind

grupo0 = df[df['Outcome'] == 0]['Glucose']
grupo1 = df[df['Outcome'] == 1]['Glucose']

ttest_ind(grupo0, grupo1, equal_var=False)
```
